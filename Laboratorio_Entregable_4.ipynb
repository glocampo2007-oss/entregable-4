```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================================================================
# Ejemplo 1: Aprendizaje Supervisado con el Dataset MNIST
# Modelo: Red Neuronal Convolucional (Clasificador)
# ==============================================================================

import tensorflow as tf
from tensorflow.keras import layers, models

print("\n--- Iniciando Ejemplo de Aprendizaje Supervisado ---")

# 1. Cargar y Preprocesar el Dataset MNIST
# El dataset MNIST contiene 60,000 imágenes de entrenamiento y 10,000 de prueba
# de dígitos escritos a mano (0-9).
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()

# Normalizar las imágenes a un rango de [0, 1]
# Las imágenes tienen un tamaño de 28x28 píxeles.
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255

# Convertir las etiquetas a formato one-hot encoding
train_labels = tf.keras.utils.to_categorical(train_labels)
test_labels = tf.keras.utils.to_categorical(test_labels)

print(f"Dimensiones de las imágenes de entrenamiento: {train_images.shape}")
print(f"Dimensiones de las etiquetas de entrenamiento: {train_labels.shape}")

# 2. Construir el Modelo de Red Neuronal Convolucional (CNN)
model_supervised = models.Sequential()
model_supervised.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
model_supervised.add(layers.MaxPooling2D((2, 2)))
model_supervised.add(layers.Conv2D(64, (3, 3), activation='relu'))
model_supervised.add(layers.MaxPooling2D((2, 2)))
model_supervised.add(layers.Conv2D(64, (3, 3), activation='relu'))
model_supervised.add(layers.Flatten())
model_supervised.add(layers.Dense(64, activation='relu'))
model_supervised.add(layers.Dense(10, activation='softmax')) # 10 clases para los dígitos 0-9

model_supervised.summary()

# 3. Compilar y Entrenar el Modelo
# Usamos 'adam' como optimizador y 'categorical_crossentropy' para la función de pérdida
# ya que las etiquetas están en one-hot encoding.
model_supervised.compile(optimizer='adam',
                         loss='categorical_crossentropy',
                         metrics=['accuracy'])

print("\nEntrenando el modelo supervisado...")
history_supervised = model_supervised.fit(train_images, train_labels, epochs=5, batch_size=64, validation_split=0.1)

# 4. Evaluar el Modelo
loss_supervised, accuracy_supervised = model_supervised.evaluate(test_images, test_labels, verbose=0)
print(f"\nPrecisión del modelo supervisado en datos de prueba: {accuracy_supervised:.4f}")

# 5. Visualización de Resultados (Supervisado)
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history_supervised.history['accuracy'], label='Precisión de entrenamiento')
plt.plot(history_supervised.history['val_accuracy'], label='Precisión de validación')
plt.xlabel('Época')
plt.ylabel('Precisión')
plt.legend()
plt.title('Precisión del Modelo Supervisado')

plt.subplot(1, 2, 2)
plt.plot(history_supervised.history['loss'], label='Pérdida de entrenamiento')
plt.plot(history_supervised.history['val_loss'], label='Pérdida de validación')
plt.xlabel('Época')
plt.ylabel('Pérdida')
plt.legend()
plt.title('Pérdida del Modelo Supervisado')
plt.show()

# Mostrar algunas predicciones del modelo
predictions_supervised = model_supervised.predict(test_images[:5])
predicted_labels_supervised = np.argmax(predictions_supervised, axis=1)
true_labels_supervised = np.argmax(test_labels[:5], axis=1)

print("\n--- Primeras 5 Predicciones (Supervisado) ---")
for i in range(5):
    plt.imshow(test_images[i].reshape(28, 28), cmap=plt.cm.binary)
    plt.title(f"Real: {true_labels_supervised[i]}, Predicho: {predicted_labels_supervised[i]}")
    plt.axis('off')
    plt.show()


# ==============================================================================
# Ejemplo 2: Aprendizaje No Supervisado con el Dataset MNIST
# Modelo: K-Means Clustering
# ==============================================================================

from sklearn.cluster import MiniBatchKMeans
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE

print("\n--- Iniciando Ejemplo de Aprendizaje No Supervisado ---")

# 1. Cargar y Preprocesar el Dataset MNIST
# Para K-Means, usaremos un subconjunto o el dataset completo aplanado.
(X_train_unsupervised, y_train_unsupervised), (_, _) = tf.keras.datasets.mnist.load_data()

# Aplanar las imágenes de 28x28 a un vector de 784 píxeles
X_train_unsupervised = X_train_unsupervised.reshape(len(X_train_unsupervised), -1).astype('float32')

# Escalar los datos para que K-Means funcione mejor
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_train_unsupervised)

print(f"Dimensiones de los datos escalados para clustering: {X_scaled.shape}")

# 2. Construir y Entrenar el Modelo K-Means
# Usaremos MiniBatchKMeans para manejar el tamaño del dataset de manera más eficiente.
# Intentamos agrupar en 10 clusters, esperando que cada cluster corresponda a un dígito.
n_digits = len(np.unique(y_train_unsupervised)) # Debería ser 10
kmeans = MiniBatchKMeans(n_clusters=n_digits, random_state=42, n_init=10, batch_size=256)

print("\nEntrenando el modelo no supervisado (K-Means). Esto puede tardar unos segundos...")
kmeans.fit(X_scaled)

# 3. Asignar etiquetas a los clusters
# No hay una correspondencia directa entre cluster_id y digit_label, así que la mapeamos.
labels_kmeans = kmeans.labels_

# Para evaluar, podemos intentar mapear los clusters a las etiquetas verdaderas.
def assign_cluster_labels(kmeans_labels, true_labels):
    predicted_labels = np.zeros_like(kmeans_labels)
    for i in range(kmeans.n_clusters):
        # Encontrar los índices de los puntos que pertenecen al cluster i
        mask = (kmeans_labels == i)
        # Encontrar la etiqueta verdadera más frecuente en ese cluster
        if np.any(mask):
            # np.bincount es robusto para arrays vacíos
            most_common_label = np.argmax(np.bincount(true_labels[mask]))
            predicted_labels[mask] = most_common_label
    return predicted_labels

predicted_labels_unsupervised = assign_cluster_labels(labels_kmeans, y_train_unsupervised)

# 4. Visualización de Resultados (No Supervisado)
from sklearn.metrics import accuracy_score
accuracy_unsupervised = accuracy_score(y_train_unsupervised, predicted_labels_unsupervised)
print(f"\nPrecisión (mapeada) del clustering K-Means: {accuracy_unsupervised:.4f}")

# Visualizar los centroides de los clusters como imágenes de dígitos
plt.figure(figsize=(10, 8))
for i in range(kmeans.n_clusters):
    ax = plt.subplot(2, 5, i + 1)
    centroid = kmeans.cluster_centers_[i].reshape(28, 28)
    # Deshacer el escalado para visualizar el centroide correctamente
    centroid_original_scale = scaler.inverse_transform(kmeans.cluster_centers_[i].reshape(1, -1)).reshape(28, 28)
    plt.imshow(centroid_original_scale, cmap='gray')
    plt.title(f'Cluster {i}')
    plt.axis('off')
plt.suptitle('Centroides de K-Means (Aprendizaje No Supervisado)', fontsize=16)
plt.show()

# Visualización de los datos reducidos con t-SNE
# Reducir la dimensionalidad para la visualización (t-SNE es computacionalmente costoso, usar un subconjunto)
print("Reduciendo dimensionalidad con t-SNE para visualización... (puede tardar)")
subset_indices = np.random.choice(len(X_scaled), 2000, replace=False)
X_subset = X_scaled[subset_indices]
y_subset = y_train_unsupervised[subset_indices]
labels_kmeans_subset = labels_kmeans[subset_indices]

tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_tsne = tsne.fit_transform(X_subset)

plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels_kmeans_subset, cmap='Paired', alpha=0.7)
plt.legend(handles=scatter.legend_elements()[0], title="Clusters")
plt.title('Visualización t-SNE de Clusters MNIST (No Supervisado)')
plt.colorbar(scatter)
plt.show()


# ==============================================================================
# Ejemplo 3: Aprendizaje por Refuerzo con el Dataset MNIST (Concepto Simplificado)
# Este es un ejemplo conceptual ya que MNIST no es un entorno RL tradicional.
# Se simula un agente que "adivina" el dígito y recibe una recompensa.
# ==============================================================================

print("\n--- Iniciando Ejemplo de Aprendizaje por Refuerzo (Concepto Simplificado) ---")

# Importar las librerías necesarias
import random
from collections import deque

# 1. Definir el "Entorno" de MNIST simplificado
class MNISTGuessingEnv:
    def __init__(self, images, labels):
        self.images = images
        self.labels = labels
        self.current_image_idx = -1
        self.action_space_n = 10 # 0-9 dígitos

    def reset(self):
        # Seleccionar una nueva imagen aleatoria del dataset
        self.current_image_idx = random.randint(0, len(self.images) - 1)
        # El estado es la imagen aplanada y normalizada
        state = self.images[self.current_image_idx].flatten() / 255.0
        return state

    def step(self, action):
        # 'action' es la predicción del agente (un dígito del 0 al 9)
        true_label = self.labels[self.current_image_idx]

        reward = 1.0 if action == true_label else 0.0
        done = True # Cada "episodio" es una única adivinanza
        info = {'true_label': true_label}
        next_state = None # No hay un "siguiente estado" en este episodio de un solo paso
        return next_state, reward, done, info

# Cargar el dataset MNIST para el entorno RL
(rl_images, rl_labels), (_, _) = tf.keras.datasets.mnist.load_data()

# Crear el entorno
env = MNISTGuessingEnv(rl_images, rl_labels)

print(f"El entorno tiene {len(rl_images)} imágenes. Espacio de acciones: {env.action_space_n}")

# 2. Definir el Agente (Simple Red Neuronal para Política/Q-values)
# Usaremos una red neuronal para estimar los Q-values (acciones-valor) o la política.
# Para simplificar, una red que predice un Q-value para cada acción.

class DQNAgent:
    def __init__(self, state_size, action_size):
        self.state_size = state_size
        self.action_size = action_size
        self.memory = deque(maxlen=2000) # Experiencia replay
        self.gamma = 0.95    # Factor de descuento
        self.epsilon = 1.0   # Exploración vs. explotación
        self.epsilon_min = 0.01
        self.epsilon_decay = 0.995
        self.learning_rate = 0.001
        self.model = self._build_model()

    def _build_model(self):
        # Red neuronal simple para aproximar Q-values
        model = models.Sequential()
        model.add(layers.Dense(64, input_dim=self.state_size, activation='relu'))
        model.add(layers.Dense(64, activation='relu'))
        model.add(layers.Dense(self.action_size, activation='linear')) # Salida: Q-value para cada acción
        model.compile(loss='mse', optimizer=tf.keras.optimizers.Adam(learning_rate=self.learning_rate))
        return model

    def remember(self, state, action, reward, next_state, done):
        self.memory.append((state, action, reward, next_state, done))

    def act(self, state):
        if np.random.rand() <= self.epsilon:
            return random.randrange(self.action_size) # Exploración aleatoria
        # Explotación: elegir la acción con el Q-value más alto
        state = np.reshape(state, [1, self.state_size]) # Añadir dimensión de batch
        act_values = self.model.predict(state, verbose=0)
        return np.argmax(act_values[0])

    def replay(self, batch_size):
        if len(self.memory) < batch_size:
            return
        minibatch = random.sample(self.memory, batch_size)
        for state, action, reward, next_state, done in minibatch:
            target = reward
            if not done: # No hay next_state en este entorno, siempre done
                pass # target ya es reward

            state = np.reshape(state, [1, self.state_size])
            target_f = self.model.predict(state, verbose=0)
            target_f[0][action] = target # Actualizar el Q-value para la acción tomada
            self.model.fit(state, target_f, epochs=1, verbose=0)

        if self.epsilon > self.epsilon_min:
            self.epsilon *= self.epsilon_decay

# Crear el agente
state_size = rl_images.shape[1] * rl_images.shape[2] # 28 * 28 = 784
action_size = env.action_space_n # 10
agent = DQNAgent(state_size, action_size)

# 3. Entrenar el Agente (Bucle de Aprendizaje por Refuerzo)
epochs_rl = 1000 # Número de "episodios" de adivinanza
batch_size_rl = 32
rewards_per_epoch = []

print("\nEntrenando el agente de Aprendizaje por Refuerzo... (esto puede tardar)")
for e in range(epochs_rl):
    state = env.reset()
    action = agent.act(state)
    next_state, reward, done, info = env.step(action)

    agent.remember(state, action, reward, next_state, done)
    agent.replay(batch_size_rl)

    rewards_per_epoch.append(reward)

    if (e + 1) % 100 == 0:
        avg_reward = np.mean(rewards_per_epoch[-100:])
        print(f"Época: {e + 1}/{epochs_rl}, Recompensa Promedio (últimos 100): {avg_reward:.2f}, Epsilon: {agent.epsilon:.2f}")

print("Entrenamiento de RL completado.")

# 4. Visualización de Resultados (Aprendizaje por Refuerzo)
plt.figure(figsize=(10, 6))
plt.plot(rewards_per_epoch)
plt.xlabel('Episodio')
plt.ylabel('Recompensa Instantánea')
plt.title('Recompensa por Episodio (Aprendizaje por Refuerzo Simplificado)')
plt.grid(True)
plt.show()

# Visualizar la recompensa promedio a lo largo del tiempo
moving_avg_rewards = np.convolve(rewards_per_epoch, np.ones(50)/50, mode='valid')
plt.figure(figsize=(10, 6))
plt.plot(moving_avg_rewards)
plt.xlabel('Episodio')
plt.ylabel('Recompensa Promedio (Ventana de 50 Episodios)')
plt.title('Recompensa Promedio Suavizada (Aprendizaje por Refuerzo Simplificado)')
plt.grid(True)
plt.show()

# Probar el agente en algunas imágenes nuevas (simuladas)
print("\n--- Probando el Agente de RL --- ")
correct_predictions_rl = 0
num_tests_rl = 100
for _ in range(num_tests_rl):
    state = env.reset()
    # Forzar la explotación para probar
    action = np.argmax(agent.model.predict(np.reshape(state, [1, agent.state_size]), verbose=0)[0])
    _, reward, _, info = env.step(action)
    if reward == 1.0:
        correct_predictions_rl += 1

accuracy_rl = correct_predictions_rl / num_tests_rl
print(f"Precisión del agente de RL (en {num_tests_rl} pruebas): {accuracy_rl:.2f}")

print("\n--- Fin de todos los ejemplos --- ")
```
